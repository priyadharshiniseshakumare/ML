```python
import pandas as pd
```

```python
df = pd.read_csv("/content/titanic.csv")
df.head()
```

```python
df.head()
df.columns
```

```python
df.drop(['Ticket', 'PassengerId'],
        axis='columns',
        inplace=True)
```

```python
X = df.drop('Survived', axis=1)
y = df['Survived']
```

```python
X.head()
```

```python
dummies = pd.get_dummies(X['Sex'])
```

```python
X = pd.concat([X,dummies], axis = 'columns')
```

```python
X.drop('Sex', axis=1, inplace=True)
```

```python
X
```

```python
X.isnull().any()
```

```python
X['Age'] =X['Age'].fillna(X['Age'].mean())
```

```python
X.isnull().any()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
```

```python
X_train_processed = X_train.drop(['Name', 'Cabin', 'Embarked', 'Ticket', 'PassengerId'], axis=1)
X_train_processed['female'] = X_train_processed['female'].astype(int)
X_train_processed['male'] = X_train_processed['male'].astype(int)

model.fit(X_train_processed,y_train)
```

```python
X_test_processed = X_test.drop(['Name', 'Cabin', 'Embarked', 'Ticket', 'PassengerId'], axis=1)
X_test_processed['female'] = X_test_processed['female'].astype(int)
X_test_processed['male'] = X_test_processed['male'].astype(int)

model.score(X_test_processed, y_test)
```

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_processed)
print(classification_report(y_test, y_pred))
```
